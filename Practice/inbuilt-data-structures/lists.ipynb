```python
#Accessing the list items
fruits = ["Apple", "banana", "Cherry", "Mango"]
print(fruits)
print("My favourite fruit",fruits[1])
print(fruits[-1])
```

```python
#Modifying the list elements
fruits[1] = "Guava"
print(fruits)
```

```python
#Jump steps
print(fruits[::2]) #jump size of 2
print(fruits[::-1])#reversing the step size
```

```python
for index,fruit in enumerate(fruits):
    print(index, fruit)
```
