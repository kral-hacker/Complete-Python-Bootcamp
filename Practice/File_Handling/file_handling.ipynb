```python
with open("example.txt", mode = 'r') as file:
    content = file.read()

print(content)
```

```python
# reading the new file line by line
with open("example.txt", mode='r') as file:
    for line in file:
        print(line.strip())
```

```python
# writing the file (overwriting)
with open("example.txt", mode='w') as file:
    file.write("hello world")
```

```python
# opening the file in the append mode
with open("example.txt", mode = 'a') as file:
    file.write("\nMy name is Ananay Tyagio")
```

```python
#writing the multiple lines into the file
lines = ["my name is ananay tyagi\n", "my age is 22"]
with open("example.txt", mode= 'a') as file:
    file.writelines(lines)
```

```python
#writing and then reading the file
with open("example1.txt", mode = 'w+') as file:
    file.write("My name is ananay tyagi")
    file.write("I am also known as kral hacker")

    file.seek(0) # moving the cursor of the file to the 0 index

    content = file.read()

print(content)
```
