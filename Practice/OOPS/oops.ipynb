```python
## constructor
class dog:
    def __init__(self, name:str, breed:str, age:int):
        self.name = name # self tells that the variable is of taht class
        self.breed = breed
        self.age = age


dog1 = dog(1, "Labra", "12")
print("The name of the dog1 is", dog1.name)
```

```python
# instance methods
class dog:
    def __init__(self, name:str, breed:str, age:int):
            self.name = name # self tells that the variable is of taht class
            self.breed = breed
            self.age = age
# this is nothing but the instance method
    def barking(self):
          print("Woof")

dog2 = dog("Simba", "labra", 10)
dog2.barking()
```
