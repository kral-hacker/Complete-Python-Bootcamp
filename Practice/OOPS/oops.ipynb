# Inheritance 

## Inheritance is one of the pillars of OOPS 

### It is when the attributes and methods of one class can be inherited by the another class    

#### Mechanism of deriving a new class from the old class is called inheritance

```python
## Inheritance
# This is the Parent class
class Car:
    def __init__(self, windows, enginetype, power):
        self.windows = windows
        self.enginetype = enginetype
        self.power = power
    def drive(self):
        print(f"The person drives the {self.power}cc car")
         
```

```python
car1 = Car(4, "petrol", 1500)
car1.drive()
```

```python
## Making the child class
## this example is of single inheritance
class Merc(Car):
    def __init__(self, windows, enginetype, power, selfdrive):
        super().__init__(windows, enginetype, power) ## Super is used to define that whether the child class will inherit from the parent class
        self.selfdrive = selfdrive
    def feature(self):
        print("This car is self driving: ",self.selfdrive)
```

```python
car2 = Merc(2, "diesel", 3400 , True)
car2.feature()
```

## Multiple Inheritance

### When a class inherits from more than 1 base class

```python
## one base class

class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        pass

## another base class
class Pet:
    def __init__(self, owner):
        self.owner = owner

    def speak(self):
        pass

class Dog(Animal, Pet):
    def __init__(self, name , owner):
        Animal.__init__(self, name)
        Pet.__init__(self, owner)

    def speak(self):
        print("The dog says woof")
```

```python
Simba = Dog("Simba", "Ananay")
Simba.speak()
```
