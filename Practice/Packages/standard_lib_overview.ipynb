```python
import json
api_dumps = {
    'name':'Ananay',
    'age':22
}
json_dumps =json.dumps(api_dumps)
print(json.dumps(api_dumps)) # to convert into the json dumps
print(json.loads(json_dumps)) # to convert json back into the dict
```

```python
import csv 

with open("examples.csv", mode = 'w', newline = "") as f:
    writer= csv.writer(f)
    writer.writerow(['name','age'])
    writer.writerow(['Ananay',22])

with open("examples.csv", mode = 'r') as f:
    reader = csv.reader(f)
    for row in reader:
        print(row)
```
