# Exception Handling is the way of handling errors gracefully and taking the corrective actions without stoppinxg the execution of the program

```python
try:
    a=b
except:
    print("The variable is not defined")
```

```python
try:
    a=b
except NameError as exg:
    print(exg)
```

```python
try:
    result  = 1/0
except ZeroDivisionError as z:
    print(z)
```

```python
try:
    result = 1/0
except ZeroDivisionError as e:
    print(e)

```

## The use of else

```python
try:
    num = int(input("Enter the number u would love to divide: "))
    result = 1/num
except ZeroDivisionError as e:
    print(e)
# when the except block is executed the program comes out of the try and except block
except ValueError as v:
    print("That's not a valid number")
# else block will always be executed after the try block cames up with no error
else:
    print(f"The result is {result}")
```

## the use of finally
It will alays be executed whether there is a error or not

```python
try:
    num = int(input("Enter the number u would love to divide: "))
    result = 1/num
except ZeroDivisionError as e:
    print(e)
except ValueError as v:
    print("That's not a valid number")
else:
    print(f"The result is {result}")
# this block always gets executed whether the error comes or not
finally:
    print("Execution completed")
```
