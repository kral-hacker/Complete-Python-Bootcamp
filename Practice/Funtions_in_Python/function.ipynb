```python
def sout(*args):
    for i in args:
        print(i,end=" ")

sout(1,2,4,5,6,7,8,"Ananay")
```

```python
def sout(*args, **rdict):
    for i in args:
        print(i,end=" ")
    for i,j in rdict.items():
        print(i," ",j)
sout(1,2,3,5,6,7,name="Ananay",lastname = "Tyagi")
```

```python
def covert_temperature(temp, unit):
    if (unit == "C"):
        return temp*9/5+32
    else:
        return temp-32*5/9
```

```python
covert_temperature(50, "C")
```

```python
# password checker

def password_checker(password):
    if len(password)<8:
        return False
    if not any(char.isdigit() for char in password):
        return False
    if not any(char.islower() for char in password):
        return False
    if all(char.isupper() for char in password):
        return False
    return True

password_checker("#Ananay")
```

## Filter Function is used to filter out elements from an iterable data structure

```python
lst = [1,2,3,4,5,6,7,8,9]

list(filter(lambda x : x%2==0, lst))
```
