Lambda Function are a small anonymous function
So basically functions withot a name 
They are defined with the lambda keyword
They can take many number of arguments but they return only one expression

```python
# lambda arguments : expression
```

```python
def addition(a,b):
    return a+b

addition(2,3)
```

```python
#Creating a lambda function

addition = lambda a,b : a+b
type(addition)

addition(2,3) # this the lambda funtion
```

```python
iseven = lambda num:bool(not(num%2))
iseven(2)
```
