## Map functions are the type of functions that apply the a particular function on every element of the collection

```python
numbers = [1,3,4,5,6,7,78]
list(map(lambda x : x*x, numbers))
```

### Map Multiple iterables

```python
numbers1 = [1,2,3]
numbers2 = [11,12,13]
list(map(lambda x,y:x+y, numbers1, numbers2))
```

### Convert the list of strings into intgers

```python
num = ["1","2","3"]
ans = list(map(lambda x: int(x), num))
print(type(num[0]))
print(type(ans[0]))
```

### Converting lower case letters into upper case letters

```python
words = ['apple', 'banana', 'cherry']
list(map(str.upper, words))
```

### Getting the dict items from the list of dict

```python
d = [
    {
        'name':'Ananay',
        'age':32
    },
    {
        'name':'Dinesh',
        'age':60
    }
]
print(d[0]['name'])
def get_name(person):
    return person['name']
list(map(get_name, d))
```
